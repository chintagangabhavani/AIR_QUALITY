```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('AIR_QUALITY.csv')

# --- DATA PREPARATION & AGGREGATION ---
# Filter data for 'city' location
city_center_data = df[df['Location'] == 'city']

# Group data by location to find average pollution levels
avg_pollution = df.groupby('Location')['PM2.5_Level'].mean().reset_index()
display(avg_pollution)

# --- THE 5 REQUIRED CHARTS ---
# Chart 1: Line Chart (Tracking PM2.5_Level over time at City Center)
plt.figure(figsize=(8, 4))
plt.plot(city_center_data['Date'], city_center_data['PM2.5_Level'], marker='o', color='red')
plt.title('Chart 1: PM2.5 Level Trend at City Center')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ensure data is loaded if previous cells haven't been run
try:
    _ = avg_pollution
except NameError:
    df = pd.read_csv('/content/AIR_QUALITY.csv')
    avg_pollution = df.groupby('Location')['PM2.5_Lev'].mean().reset_index()

# Set 'Location' as the index so the bar labels show properly
avg_pollution_indexed = avg_pollution.set_index('Location')

# Chart 2: Bar Chart (Comparing Average PM2.5 by Location)
plt.figure(figsize=(8, 4))
avg_pollution_indexed['PM2.5_Level'].plot(kind='bar', color='orange')
plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 Levels over Dates)
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2.5_Level'], color='purple', s=50)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset with correct filename casing
df = pd.read_csv('/content/AIR_QUALITY.csv')

# Chart 4: Histogram of PM2.5 Levels
plt.figure(figsize=(8, 4))
plt.hist(df['PM2.5_Level'], bins=5, color='skyblue', edgecolor='black')
plt.title('Chart 4: Frequency of different PM2.5 level')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.show()
```

```python
 #Chart 5: Horizontal Bar Chart of Average PM2.5 by Location
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].mean().plot(kind='barh', color='teal')
plt.title('Average PM2.5 Levels by Location')
plt.xlabel('Average PM2.5')
plt.ylabel('Location')
plt.show()
```

```python
# Chart 6: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='bar', color='teal')
plt.title('Chart 5: Maximum PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Peak PM2.5 Level')
plt.xticks(rotation=0)
plt.show()
```
